```python
# Import necessary libraries
import cv2
import numpy as np
from IPython.display import display, Image
```

```python
# Load an image
image = cv2.imread('/content/image.jpeg')  # Replace 'image.jpg' with your image path
```

```python
# Display the original image
display(Image(data=cv2.imencode('.jpg', image)[1].tobytes()))
```

```python
# Convert the image to grayscale
gray_image = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)

# Display the grayscale image
display(Image(data=cv2.imencode('.jpg', gray_image)[1].tobytes()))
```

```python
# Apply Gaussian blur to reduce noise
blurred_image = cv2.GaussianBlur(gray_image, (5, 5), 0)

# Display the blurred image
display(Image(data=cv2.imencode('.jpg', blurred_image)[1].tobytes()))
```

```python
# Perform edge detection using Canny
edges = cv2.Canny(blurred_image, 50, 150)

# Display the edges
display(Image(data=cv2.imencode('.jpg', edges)[1].tobytes()))
```

```python
# Find contours in the image
contours, _ = cv2.findContours(edges, cv2.RETR_EXTERNAL, cv2.CHAIN_APPROX_SIMPLE)

# Draw contours on the original image
cv2.drawContours(image, contours, -1, (255, 0, 0), 2)

# Display the image with contours
display(Image(data=cv2.imencode('.jpg', image)[1].tobytes()))
```
